```python
plt.figure(figsize=(5,6))
sns.countplot(x='month', data=ds)
plt.xlabel('Month')
plt.ylabel('Frequency')
plt.title('Month Count Plot')
plt.show()
```

```python
freq = ds['month'].value_counts()

plt.figure(figsize=(4,6))
plt.pie(freq, labels=freq.index, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
freq = ds['month'].value_counts()

plt.figure(figsize=(5,6))
plt.bar(freq.index, freq.values)
plt.xlabel('Month')
plt.ylabel('Frequency')
plt.title("Bar Chart")
plt.show()
```

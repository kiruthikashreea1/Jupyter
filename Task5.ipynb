```python
#line plot 
plt.figure(figsize=(5,3))
plt.plot(DS['student_id'],DS['cgpa'])
plt.xlabel("student_id")
plt.ylabel("CGPA")
plt.title("student_id vs CGPA")
plt.show()
```

```python
#Trends using if statement
value1=DS["cgpa"].iloc[0]
value2=DS["cgpa"].iloc[-4]

if value2 > value1:
    print("Increasing Trends")
elif value2 < value1:
    print("Decreasing Trends")
else:
    print("Stable Trends")
```

```python
# IQR
Q1 = numeric.quantile(0.25)
Q3 = numeric.quantile(0.75)

IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
Upper_limit = Q3 + 1.5 * IQR

outliers = (
    (numeric < lower_limit) | (numeric > Upper_limit)
)
print("Lower Bound:\n", lower_limit)
print("Upper Bound:\n", Upper_limit)
print("Outlier in each column:\n", numeric[outliers].count())
```

```python
# Cv
mean_values = numeric.mean()
std_values = numeric.std()

result6 = pd.DataFrame({
    "Mean": mean_values,
    "Standard Deviation": std_values,
    "CV (%)": (std_values / mean_values) * 100
})
print(result6.round(3))
```

```python
# Compare Mean and Median to Identify Skewness
result7 = pd.DataFrame({
    "Mean": numeric.mean(),
    "Median": numeric.median()
})

result7["Skewness"] = "Symmetric"
result7.loc[result7["Mean"] > result7["Median"], "Skewness"] = "Positive Skewness"
result7.loc[result7["Mean"] < result7["Median"], "Skewness"] = "Negative Skewness"

print(result7.round(3))
```

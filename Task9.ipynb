```python
import numpy as np
import seaborn as sns
```

```python
num = Ds.select_dtypes(include=np.number)
print(num.columns)
```

```python
#SKEWNESS
column= num.columns[0]
sns.histplot(num[column],kde=True)
plt.title(f"Skewness Distribution of {column}")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()

print("Skewness:", num[column].skew())
```

```python
#skewness
result4=pd.DataFrame({
    "Skewness": num.skew()
})
print(result4.round(3))
```

```python
#Kurtosis
column= num.columns[0]
sns.histplot(num[column],kde=True)
plt.title(f" Distribution of {column}")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()

print("Kurtosis",num[column].kurt())
```

```python
result4 = pd.DataFrame({
    "kurt": num.kurt()
})

print(result4.round(3))
```

```python
# CREATE HISTOGRAM
plt.figure(figsize=(19,11))
plt.subplot(5,6,25)

plt.hist(num["lead_time"].dropna(), bins=50, edgecolor="pink")

plt.title("Histogram of Lead Time")
plt.xlabel("Lead Time")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))
plt.subplot(5,6,25)

plt.hist(num["total_of_special_requests"].dropna(), bins=50, edgecolor="pink")

plt.title("Histogram of Total Special Requests")
plt.xlabel("Total Special Requests")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# BOX PLOT
plt.figure(figsize=(12,8))
plt.subplot(2,2,1)

sns.boxplot(y=num["lead_time"])
plt.title("Box Plot of Lead Time")
plt.ylabel("Lead Time")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))
plt.subplot(2,2,3)

sns.boxplot(y=num["stays_in_week_nights"])
plt.title("Box Plot of Stays in Week Nights")
plt.ylabel("Stays in Week Nights")

plt.tight_layout()
plt.show()
```

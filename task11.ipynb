```python
ds['country'] = ds['country'].fillna(ds['country'].mode()[0])
ds['days_in_waiting_list'] = ds['days_in_waiting_list'].fillna(ds['days_in_waiting_list'].mean())
```

```python
ds.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder

lble = LabelEncoder()

ds['hotel_encoded'] = lble.fit_transform(ds['hotel'])
```

```python
ds.head()
```

```python
#SCATTER
sns.scatterplot(data=ds, x='total_of_special_requests', y='adults')
plt.title('Scatter Plot')
plt.xlabel('Lead Time')
plt.ylabel('Average Daily Rate')
plt.show()
```

```python
sns.violinplot(data=ds, x='hotel', y='adr', hue='hotel')
plt.title("Violin Plot")
plt.show()
```

```python
sns.boxplot(data=ds, x='hotel', y='adults', hue='hotel')
plt.title("Box Plot")
plt.show()
```

```python
sns.barplot(data=ds, x='hotel', y='adr', hue='hotel')
plt.title("Bar Plot")
plt.show()
```

```python
print("Skewness of ADR:", ds['adr'].skew())
```

```python
corr= ds[['lead_time', 'adr', 'adults', 'total_of_special_requests']].corr()
print(corr)
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
